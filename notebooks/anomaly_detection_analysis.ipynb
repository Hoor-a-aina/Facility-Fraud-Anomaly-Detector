```python
raw = generate_transactions()
data = add_behavioural_features(raw)
print('Rows:', len(data))
print('Suspicious:', int(data.is_suspicious.sum()))
print('Positive rate:', round(data.is_suspicious.mean() * 100, 3), '%')
display(data.head())
assert len(data) >= 1400 and data.is_suspicious.mean() < 0.02
```

## 2. Why accuracy is not enough
The all-normal baseline can look strong because almost every record is normal. PR-AUC and Precision@50 better reveal whether high-ranked alerts contain rare suspicious examples.

```python
normal_baseline_accuracy = (data.is_suspicious == 0).mean()
print(f'All-normal baseline accuracy: {normal_baseline_accuracy:.3%}')
print('But this baseline detects 0 suspicious events.')
```
